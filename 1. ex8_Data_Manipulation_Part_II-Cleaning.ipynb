```python
# Q1-Q4: Load, inspect, check Permit Number duplicates, rename columns
df = pd.read_csv("data/Building_Permits.csv", low_memory=False)
df_clean = df.copy()
print(df_clean.head())
df_clean.info()
print(df_clean["Permit Number"].duplicated().value_counts())
df_clean = df_clean.rename(columns={"Permit Creation Date": "Date Created", "Street Number": "House Number"})
```

```python
# Q5-Q8: Missing percentage, Street Number Suffix & Zipcode nulls, dropna rows/columns
missing_pct = (df_clean.isna().sum().sum() / df_clean.size) * 100
print(f"Missing percentage: {missing_pct:.2f}%")
print("Street Number Suffix nulls:", df_clean["Street Number Suffix"].isna().sum())
print("Zipcode nulls:", df_clean["Zipcode"].isna().sum())
print("Drop rows any NA:", df_clean.dropna(how="any").shape)
print("Drop cols any NA:", df_clean.dropna(how="any", axis=1).shape)
print("Drop cols all NA:", df_clean.dropna(how="all", axis=1).shape)
```

```python
# Q9-Q10: Backfill then fill 0, and remove rows where Permit Type Definition contains sign - erect
df_filled = df_clean.bfill().fillna(0)
df_filtered = df_clean[~df_clean["Permit Type Definition"].fillna("").str.contains("sign - erect", case=False, regex=False)].copy()
print("Remaining NAs:", df_filled.isna().sum().sum(), "| Filtered shape:", df_filtered.shape)
```
