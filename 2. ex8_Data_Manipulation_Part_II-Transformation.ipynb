```python
# Q1-Q6: Load movies.csv, inspect, deduplicate, check nulls, rename RunTime/Gross
import re
m_df = pd.read_csv("data/movies.csv")
print(m_df.head())
m_df.info()
print("Exact duplicates:", m_df.duplicated().sum())
m_df = m_df.drop_duplicates()
print("Duplicate MOVIES:", m_df.duplicated(subset=["MOVIES"]).sum())
m_df = m_df.drop_duplicates(subset=["MOVIES"])
print(m_df.isna().sum())
m_df = m_df.rename(columns={"RunTime": "RUNTIME", "Gross": "GROSS"})
```

```python
# Q7-Q8: Clean text and extract DIRECTORS and STARS
m_df["GENRE"] = m_df["GENRE"].map(lambda s: s.replace("\n", "").strip() if isinstance(s, str) else s)
m_df["ONE-LINE"] = m_df["ONE-LINE"].transform(lambda s: s.replace("\n", "").strip() if isinstance(s, str) else s)
m_df["STARS"] = m_df["STARS"].apply(lambda s: s.replace("\n", "").strip() if isinstance(s, str) else s)

def extract_directors(val):
    if not isinstance(val, str):
        return ""
    first_part = val.split("|")[0].strip()
    if re.match(r"^(Director|Directors):", first_part, flags=re.IGNORECASE):
        return first_part.split(":", 1)[1].strip()
    return ""

def extract_stars(val):
    if not isinstance(val, str):
        return ""
    for part in [p.strip() for p in val.split("|")]:
        if re.match(r"^(Star|Stars):", part, flags=re.IGNORECASE):
            return part.split(":", 1)[1].strip()
    return "" if re.match(r"^(Director|Directors):", val.strip(), flags=re.IGNORECASE) else val.strip()

m_df["DIRECTORS"] = m_df["STARS"].apply(extract_directors)
m_df["STARS"] = m_df["STARS"].apply(extract_stars)
print(m_df[["MOVIES", "DIRECTORS", "STARS"]].head())
```

```python
# Q9-Q14: Cast GROSS, RUNTIME, RATING, VOTES and standardise YEAR
m_df["GROSS"] = pd.to_numeric(m_df["GROSS"].astype(str).str.replace(r"[\$,]", "", regex=True), errors="coerce").fillna(0.0).astype(float)
m_df["RUNTIME"] = pd.to_numeric(m_df["RUNTIME"], errors="coerce")
m_df["RUNTIME"] = m_df["RUNTIME"].fillna(m_df["RUNTIME"].mean())
m_df["RATING"] = pd.to_numeric(m_df["RATING"], errors="coerce")
m_df["RATING"] = m_df["RATING"].fillna(m_df["RATING"].median())
m_df["VOTES"] = pd.to_numeric(m_df["VOTES"].astype(str).str.replace(",", "", regex=False), errors="coerce")
m_df["VOTES"] = m_df["VOTES"].fillna((m_df["VOTES"].min() + m_df["VOTES"].max()) / 2).round().astype(int)

def standardise_year(val):
    if pd.isna(val):
        return "NA"
    text = str(val).strip()
    if not text or text.lower() == "nan":
        return "NA"
    years = re.findall(r"\b\d{4}\b", text)
    if len(years) >= 2:
        return f"{years[0]}-{years[1]}"
    if len(years) == 1:
        y = years[0]
        if re.search(rf"\b{y}\s*[-–]", text):
            return f"{y}-"
        if re.search(rf"[-–]\s*\b{y}\b", text):
            return f"-{y}"
        return y
    return "NA"

m_df["YEAR"] = m_df["YEAR"].apply(standardise_year)
print(m_df["YEAR"].unique()[:20])
```
