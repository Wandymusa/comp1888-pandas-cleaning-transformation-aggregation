```python
# Products: apply() vs transform() for Purchase_mean
product = pd.read_csv("data/products.csv")
prod_copy = product.copy()
mean_purchase = product.groupby("User_ID")["Purchase"].apply("mean").rename("Purchase_mean").reset_index()
product = product.merge(mean_purchase, on="User_ID", how="left")
prod_copy["Purchase_mean"] = prod_copy.groupby("User_ID")["Purchase"].transform("mean")
print("Max diff:", (product["Purchase_mean"] - prod_copy["Purchase_mean"]).abs().max())
```

```python
# F500 Q1-Q8: Profits, sectors, top revenue, country/sector, rank gains, profit prediction
companies = pd.read_csv("data/f500.csv")
print("Max profit:", companies.loc[companies["profits"].idxmax(), "company"])
print("Min profit:", companies.loc[companies["profits"].idxmin(), "company"])
sector_counts = companies.groupby("sector")["industry"].count().sort_values(ascending=True)
print(sector_counts)
top_sector = sector_counts.index[-1]
print("Top sector:", top_sector)
top_sector_df = companies[companies["sector"] == top_sector]
print("Highest revenue in top sector:", top_sector_df.loc[top_sector_df["revenues"].idxmax(), "company"])
print(companies.groupby(["country", "sector"])["company"].count().sort_values(ascending=False).head(5))
print("Improved rank count:", len(companies[companies["rank"] < companies["previous_rank"]]))
rate = companies["profit_change"] / companies["profits"]
companies["predicted_profit"] = companies["profits"] * (1 + rate)
print(companies[["company", "profits", "profit_change", "predicted_profit"]].head())
```
